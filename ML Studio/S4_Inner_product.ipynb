# Understanding Vectors, Lengths, and Angles

In this snippet, we explore the concept of **vectors** in a 2D space, how to calculate their **lengths**, and the **angle** between them using **inner products (dot products)**. Here’s a breakdown of what we learn:

#### Vectors
A **vector** is a list of numbers where the order matters, it corresponds to a mathematical object that has both a **magnitude** (length) and a **direction**. In this example, we work with two vectors:

- $\mathbf{v_1} = [1, 0]$ (a vector pointing along the x-axis)
- $\mathbf{v_2} = [-1, 0]$ (a vector pointing in the opposite direction along the x-axis)


````python
import numpy as np
import matplotlib.pyplot as plt

# Define two 2D vectors
v1 = np.array([1, 0])  # Example vector 1
v2 = np.array([-1, 1])  # Example vector 2

# Plot the vectors
fig, ax = plt.subplots(figsize=(6, 6))
ax.set_xlim(-5, 5)
ax.set_ylim(-5, 5)

# Draw vectors
ax.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1, color='r', label='Vector 1')
ax.quiver(0, 0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='b', label='Vector 2')

# Add grid and labels
ax.grid(True, linestyle='--', linewidth=0.5)
ax.axhline(0, color='black', linewidth=1)
ax.axvline(0, color='black', linewidth=1)
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_title('2D Vectors and Inner Product')
ax.legend()

plt.show()
````

#### Length of Vectors (Magnitude)
The **length** (or **magnitude**) of a vector $ \mathbf{v} = [v_1, v_2] $ in a 2D space is given by the **Euclidean norm**:

$$
\text{Length of } \mathbf{v} = \| \mathbf{v} \| = \sqrt{v_1^2 + v_2^2}
$$

For example:
- The length of vector $ \mathbf{v_1}$  is $$ \| \mathbf{v_1} \| = \sqrt{1^2 + 0^2} = 1 $$.
- The length of vector $ \mathbf{v_2} $ is $$ \| \mathbf{v_2} \| = \sqrt{(-1)^2 + 0^2} = 1 $$.


````python
import numpy as np

# Define two 2D vectors
v1 = np.array([1, 0])  # Example vector 1
v2 = 1*np.array([-1, 0])  # Example vector 2

# Compute magnitudes (lengths) of the vectors
length_v1 = np.linalg.norm(v1)
length_v2 = np.linalg.norm(v2)

# Print lengths
print(f"Vector 1: {v1}, Length: {length_v1:.2f}")
print(f"Vector 2: {v2}, Length: {length_v2:.2f}")
````

#### Inner Product (Dot Product)
The **inner product** (or **dot product**) of two vectors $ \mathbf{v_1} = [v_{1x}, v_{1y}] $ and $ \mathbf{v_2} = [v_{2x}, v_{2y}]$ is calculated as:

$$
\mathbf{v_1} \cdot \mathbf{v_2} = (v_{1x} \times v_{2x}) + (v_{1y} \times v_{2y})
$$

For our vectors:
$$
\mathbf{v_1} \cdot \mathbf{v_2} = (1 \times -1) + (0 \times 0) = -1
$$

This inner product tells us about the **relationship** between the two vectors. If the inner product is 0, the vectors are **perpendicular**. If the inner product is positive, the vectors point in a similar direction, and if negative, they point in opposite directions.

````python
# Compute inner product (dot product)
inner_product = np.dot(v1, v2)

# Print inner product result
print(f"Inner product is: ({v1[0]} * {v2[0]}) + ({v1[1]} * {v2[1]}) = {v1[0] * v2[0]} + {v1[1] * v2[1]} = {inner_product}")
````


#### Angle Between Vectors
The **cosine** of the angle \( \theta \) between two vectors is given by the formula:

$$
\cos(\theta) = \frac{\mathbf{v_1} \cdot \mathbf{v_2}}{\|\mathbf{v_1}\| \|\mathbf{v_2}\|}
$$

For our vectors:
$$
\cos(\theta) = \frac{-1}{1 \times 1} = -1
$$


This tells us that the two vectors are **antiparallel**, meaning they point in opposite directions.

````python
# Compute the cosine of the angle between the vectors
cos_theta = inner_product / (length_v1 * length_v2)
theta = np.arccos(cos_theta)  # Compute theta in radians
theta_degrees = np.degrees(theta)  # Convert to degrees

# Print angle in degrees
print(f"Cosine of the angle: {cos_theta:.2f}")
print(f"Angle θ: {theta_degrees:.2f} degrees")
````

## Matrix Multiplication

### Understanding Matrices as Linear Transformations

Now we explore how a **matrix transforms vectors** in 2D space.  

We will see how a matrix transforms the vectors by rotating and scaling them.

### Matrices rotate and scale the data

First let usshow matrices that only rotate the data.
We create some synthetic data and rotate it!

Note the location of the origin.

````python
from sklearn.datasets import make_blobs

# Create simple dataset
X, y = make_blobs(n_samples=200, centers=2, random_state=42)
X= X - X.mean(axis=0)
# Plot original data
plt.figure(figsize=(5,5))
plt.scatter(X[:,0], X[:,1], c=y)
plt.title("Original data")
plt.grid(True)
plt.show()
````

### Step 1: Rotate the data

We apply a special transformation that rotates the data. You can play with how much rotation you want.


````python
# Distortion matrix (stretch x-axis)
rotation= np.pi/2 
A = np.array([[np.cos(rotation), -np.sin(rotation)],
              [np.sin(rotation), np.cos(rotation)]])

X_rot = X @ A.T

plt.figure(figsize=(5,5))
plt.scatter(X[:,0], X[:,1], c='gray')
plt.scatter(X_rot[:,0], X_rot[:,1], c=y)
plt.title("Rotated data (one feature dominates)")
plt.grid(True)
plt.show()
````

We can rotate the data back!

````python
# Distortion matrix (stretch x-axis)
rotation2= -rotation
A = np.array([[np.cos(rotation2), -np.sin(rotation2)],
              [np.sin(rotation2), np.cos(rotation2)]])

X_rot2 = X_rot @ A.T

plt.figure(figsize=(5,5))
plt.scatter(X[:,0], X[:,1], c='gray')
plt.scatter(X_rot2[:,0], X_rot2[:,1], c=y)
plt.title("Rotated data (one feature dominates)")
plt.grid(True)
plt.show()
````

## Step 1: Define a 2x2 transformation matrix

We use the matrix:

$$
A = \begin{bmatrix} 2 & 1 \\ 0 & 1 \end{bmatrix}
$$

This matrix will transform vectors in 2D space.

````python
import numpy as np

# Define the 2x2 matrix
A = np.array([[2, 1],
              [0, 1]])
A
````

## Step 2: Define basis vectors

In 2D, the standard basis vectors are:

$$
e_1 = \begin{bmatrix}1 \\ 0\end{bmatrix}, \quad 
e_2 = \begin{bmatrix}0 \\ 1\end{bmatrix}
$$

We want to see where the matrix sends these vectors.

````python
# Basis vectors
e1 = np.array([1, 0])
e2 = np.array([0, 1])

# Apply transformation (matrix multiplication)
Ae1 = A @ e1
Ae2 = A @ e2

Ae1, Ae2
````

## Step 3: Transform a general vector

Take a vector 

$$
v = \begin{bmatrix}3 \\ 2\end{bmatrix} = 3 e_1 + 2 e_2
$$

We apply the matrix to it:

$$
A v = 3 A e_1 + 2 A e_2
$$

````python
v = np.array([3, 2])
Av = A @ v
Av
````

\begin{align*}
A v &= 3 A e_1 + 2 A e_2 \\
     &= 3 \begin{bmatrix}2 \\ 0\end{bmatrix} + 2 \begin{bmatrix}1 \\ 1\end{bmatrix} \\
     &= \begin{bmatrix}8 \\ 2\end{bmatrix}
\end{align*}

## Step 4: Visualize the matrix transformation

We plot:
- Original basis vectors `e1`, `e2`  
- Transformed basis vectors `A*e1`, `A*e2`  
- Original vector `v`  
- Transformed vector `A*v`

````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6,6))
ax.set_xlim(-1, 9)
ax.set_ylim(-1, 6)

# Original basis
ax.quiver(0,0,e1[0],e1[1], color='blue', scale=1, angles='xy', scale_units='xy', label='e1')
ax.quiver(0,0,e2[0],e2[1], color='green', scale=1, angles='xy', scale_units='xy', label='e2')

# Transformed basis (alpha for transparency)
ax.quiver(0,0,Ae1[0],Ae1[1], color='blue', alpha=0.4, scale=1, angles='xy', scale_units='xy', label='A*e1')
ax.quiver(0,0,Ae2[0],Ae2[1], color='green', alpha=0.4, scale=1, angles='xy', scale_units='xy', label='A*e2')

# Original vector
ax.quiver(0,0,v[0],v[1], color='red', scale=1, angles='xy', scale_units='xy', label='v')

# Transformed vector
ax.quiver(0,0,Av[0],Av[1], color='red', alpha=0.4, scale=1, angles='xy', scale_units='xy', label='A*v')

ax.grid(True, linestyle='--', linewidth=0.5)
ax.axhline(0, color='black')
ax.axvline(0, color='black')
ax.set_title("Matrix as a Linear Transformation")
ax.legend()
plt.show()
````

Let’s compare people using vectors:

- Feature 1: Age  
- Feature 2: Income  

We expect people with similar age and income to be **similar**.

But what happens if one feature (income) is much larger?

## A linear classifier is just a dot product

A linear classifier computes:

\[
y = \text{sign}(w \cdot x + b)
\]

This means:
👉 It checks how much a point aligns with a direction.

````python
# Define a simple classifier
w = np.array([1, -1])
b = 0

# Grid for visualization
xx, yy = np.meshgrid(np.linspace(-5,5,100),
                     np.linspace(-5,5,100))

Z = np.sign(w[0]*xx + w[1]*yy + b)

plt.contourf(xx, yy, Z, alpha=0.3)
plt.quiver(0,0,w[0],w[1], color='red', label='w')

plt.title("Linear classifier = direction check")
plt.grid(True)
plt.legend()
plt.show()
````

## Similarity in practice: user preferences

Let’s model users as vectors:

Each column = a movie genre  
Each row = a user  

Values = how much the user likes that genre

````python
import numpy as np
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
import seaborn as sns
import matplotlib.pyplot as plt

# Users: [Action, Romance, Comedy]
X = np.array([
    [5, 0, 0],   # User 0: pure action lover
    [4, 0, 0],   # User 1: mostly action
    [0, 5, 0],   # User 2: pure romance lover
    [0, 4, 0],   # User 3: mostly romance
    [0, 0, 5],   # User 4: pure comedy lover
    [0, 0, 4],   # User 5: mostly comedy
    [3, 3, 3],   # User 6: balanced across all
    [4, 2, 1],   # User 7: action-heavy mix
    [1, 4, 2],   # User 8: romance-heavy mix
    [2, 1, 4],   # User 9: comedy-heavy mix
    [2, 2, 1],   # User 10: action+romance mix
    [1, 2, 2],   # User 11: romance+comedy mix
    [2, 1, 2],   # User 12: action+comedy mix
])

# Create DataFrame
df = pd.DataFrame(
    X,
    columns=["Action", "Romance", "Comedy"],
    index=[f"User {i}" for i in range(len(X))]
)


# Cosine similarity
cos_sim = cosine_similarity(df)

sim_df = pd.DataFrame(
    cos_sim,
    index=df.index,
    columns=df.index
)


# Heatmap
plt.figure(figsize=(10,10))
sns.heatmap(sim_df, annot=True, cmap="coolwarm", vmin=0, vmax=1)
plt.title("User similarity (cosine)")
plt.xlabel("User")
plt.ylabel("User")
plt.show()
````
