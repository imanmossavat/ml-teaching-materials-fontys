Today you will use a pretrained YOLO model to detect objects in images.

You do not need to understand how YOLO works internally. The goal is to experiment with it and investigate what it can and cannot recognize.


```python
# ============================================================
# STEP 1: INSTALL YOLO
# ============================================================

# Colab does not necessarily have the Ultralytics package
# available yet, so we install it first.

!pip install -q ultralytics
```

```python
# ============================================================
# STEP 2: LOAD YOLO
# ============================================================

from ultralytics import YOLO

# We load a pretrained YOLO model.
#
# Important:
# We are NOT training a model today.
#
# Someone has already trained this model.
# We are simply using it.

model = YOLO("yolo11n.pt")

print("YOLO model loaded!")
```

```python
# ============================================================
# STEP 3: LET YOLO LOOK AT AN IMAGE
# ============================================================

# This is an example image.
#
# We give the image to the model and ask it to
# detect objects.

# you can change this URL to any image on the web.
url = "https://ultralytics.com/images/bus.jpg"

results = model(url)

# Show the original image with YOLO's detections.
results[0].show()
```

```python
# ============================================================
# STEP 4: UPLOAD YOUR OWN IMAGE (note that this is uploaded to Google, watch for privacy of yourself and others)
# ============================================================

from google.colab import files

# Colab will show an upload button.
#
# Choose an image from your computer.

uploaded = files.upload()

# Get the name of the uploaded file.
filename = list(uploaded.keys())[0]

print("Analysing:", filename)
```

```python
# ============================================================
# STEP 5: RUN YOLO ON YOUR IMAGE
# ============================================================

# Give your image to YOLO.

results = model(filename)

# Display the predictions.

results[0].show()
```

```python
# ============================================================
# STEP 6: CHANGE THE CONFIDENCE
# ============================================================

# YOLO gives each detection a confidence value.
#
# Here we tell YOLO:
#
# "Only show detections where you are at least
#  50% confident."

results = model(
    url,
    conf=0.5
)

results[0].show()
```
